Databricks uses Apache Spark to process data
Sitting at the heart of Databricks is the engine that does this data processing: an open-source technology called Apache Spark. And this is no surprise. Spark is the dominant data processing tool in the world of big data, and Databricks was founded by the creators of Spark.

Databricks uses Apache Spark as its core engine to process big data in a fast, distributed, and parallel way. While you can run Spark yourself, managing clusters and maintaining Spark is complex and time-consuming.

Because Databricks was built by the creators of Spark, it runs an optimized, faster, and more efficient version of Spark, and then extends it into a full data platform rather than just a processing engine.

Traditional databases and data warehouses store your data in proprietary formats, meaning you can only access it through their systems and may even have to pay to read it out.

Databricks does not store your data. It reads from and writes to your own cloud storage (like S3, ADLS Gen2, or GCS) using open-source formats.

This gives you full control of your data, clear visibility of where it’s stored, and no vendor lock-in—you can access the data even without Databricks.

### 9.is databricks a SAAS,PAAS or IAAS 

- Databricks is primarily a Platform as a Service (PaaS). This is because it provides a fully managed platform where users can develop, run, and scale data engineering, analytics, and machine learning workloads without managing the underlying infrastructure. Users do not need to worry about server provisioning, Spark installation, cluster configuration, scaling, or maintenance, as all of this is handled by Databricks.
- Databricks is not Infrastructure as a Service (IaaS) because it does not require users to manage virtual machines, operating systems, or networking components. Those responsibilities are abstracted away and handled by Databricks using the cloud provider’s infrastructure.
- Databricks can be considered partially Software as a Service (SaaS) because it offers ready-to-use features such as Databricks SQL, dashboards, notebooks, and a web-based user interface. However, it is not purely SaaS since users still write code, design pipelines, and control application logic.
- In summary, Databricks sits mainly in the PaaS layer, built on top of cloud IaaS (AWS, Azure, or GCP), with some SaaS-like capabilities for analytics and visualization.

## Terminologies related to Databricks

🔹Delta lake     
delta Lake format also supports your atomicity, consistency, reliability, and durability (ACID)
Databricks have developed and released their own open-source data storage format, called Delta Lake. Delta Lake extends upon the open-source Apache Parquet storage format (which is Spark’s preferred storage format) by adding a “transaction log”, which is a list of all operations performed on your data. But the data itself remains in the well-known Parquet format, and can be accessed without using Databricks or even Spark

🔹 Data Lake     
A data lake is a centralized storage system that holds raw data in any format (structured, semi-structured, unstructured). It is highly scalable and low cost, but traditionally lacks strong governance, performance, and reliability guarantees.

🔹 Data Warehouse     
A data warehouse stores processed, structured data optimized for fast SQL queries, analytics, and reporting. It provides high performance and reliability but is usually expensive and less flexible for advanced workloads like ML.

🔹 Data Lakehouse:
A data lakehouse combines the low-cost storage and flexibility of a data lake with the performance and reliability of a data warehouse. Databricks pioneered this concept so the same data can be used for analytics, data engineering, and machine learning without duplication.

🔹 Spark + Photon     
- Apache Spark is the distributed processing engine used by Databricks for large-scale data processing.
- Photon is Databricks’ native, vectorized query engine written in C++ that accelerates SQL and Spark workloads.

👉 Together, Spark + Photon provide highly parallel, fast, and cost-efficient processing.

🔹 MLflow
MLflow is an open-source platform (created by Databricks) used to manage the machine learning lifecycle, including:
- Experiment tracking
- Model versioning
- Model deployment

It helps move ML models from experimentation to production.

🔹 Databricks SQL       
Databricks SQL is a fully managed SQL analytics service that allows users to:
- Run high-performance SQL queries
- Build dashboards
- Connect BI tools
- It brings warehouse-like analytics to the lakehouse.

🔹 Delta Live Tables (DLT)     
Delta Live Tables is a managed framework for building reliable, declarative data pipelines.

You define what the pipeline should do, and Databricks handles:
- Data quality checks
- Pipeline orchestration
- Monitoring and recovery
- This reduces manual ETL complexity.

🔹 Delta Lake         
Delta Lake is an open-source storage format developed by Databricks.

Key points:
- Built on Apache Parquet
- Adds a transaction log that records all changes
- Supports ACID properties (Atomicity, Consistency, Isolation, Durability)

👉 Even though Delta Lake adds reliability, the actual data remains in Parquet, meaning:
- Data is not locked in
- It can be accessed without Databricks or Spark

🔹 Apache Parquet
What it is:

- Apache Parquet is an open-source, columnar storage file format for storing large-scale data efficiently.
- It is widely used in big data processing systems, including Apache Spark, Hive, Presto, and Databricks.

Key Features:

Columnar storage

- Stores data column-wise instead of row-wise.
- Efficient for analytical queries that only read some columns.

Efficient storage & performance

- Supports compression and encoding, reducing disk usage.
- Speeds up queries by reading only needed columns.

Schema support

- Stores schema metadata with the data, making it self-describing.
- Ensures compatibility and data evolution over time.

Open source & cross-platform

- Can be read/written by multiple tools, not tied to one vendor.
- Works with Delta Lake, Spark, Hive, and other analytics engines.

Why Databricks uses it:

Delta Lake is built on Parquet and adds a transaction log (ACID support).

Using Parquet ensures:

- High performance
- Open format
- No vendor lock-in
- Data remains accessible even outside Databricks.

> Apache Parquet is an open-source columnar storage format that efficiently stores large-scale data and is the foundation for Delta Lake in Databricks.

-- Why Databricks vs Pandas/Hadoop?                                     
1️⃣ Hadoop
- Think of Hadoop as an old-school, giant data warehouse.
- Stores huge amounts of data across many computers (distributed).
- Processes data using MapReduce (split the task → compute in parallel → combine results).
- Good for: Offline, batch processing of very large datasets.
- Limitation: Slow for iterative or real-time tasks.
- Example: Counting all Netflix views for the last 10 years.

2️⃣ Spark
- Spark is a faster, modern version of Hadoop.
- Uses in-memory processing → no need to read/write from disk every time.
- Handles batch, streaming, SQL, and ML tasks.
- Good for: Real-time analytics and iterativecomputations.
- Example: Finding trending Netflix shows every day in seconds.

3️⃣ Databricks

- Databricks is a managed platform built on Spark.
- You don’t need to worry about:
  * Installing Spark
  * Managing clusters
  * Scaling up/down
- Provides notebooks, collaborative workspace, and ML tools.
- Integrates Delta Lake, Photon, MLflow, and SQL analytics.
- Example: Netflix or Shell using Databricks to analyze user data + run ML models without managing infrastructure.

4️⃣ Snowflake
- Snowflake is a cloud data warehouse.
- Stores clean, structured data optimized for SQL queries and analytics.
- Good for: BI dashboards and reporting.
- Limitation: Not designed for unstructured data or AI/ML pipelines.
- Example: Checking how many users watched a particular show last month.

5️⃣ Photon

- Photon is Databricks’ high-performance engine written in C++.
- Makes SQL queries and Spark jobs faster.
- Works under the hood, you don’t write code for it.
- Example: Reports that took 10 minutes now run in 10 seconds.

6️⃣ Pandas

- Pandas is a Python library for data manipulation.
- Works in-memory on a single machine.
- Good for: Small datasets, prototyping, or learning.
- Limitation: Can’t handle massive datasets.
- Example: Counting grades of students in your college class spreadsheet.
