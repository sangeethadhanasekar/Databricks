## 1- Spark architecture (driver, executors, DAG)

[medium blog](https://medium.com/@amitjoshi7/spark-architecture-a-deep-dive-2480ef45f0be)

1️⃣ Driver

- Central control program of a Spark application
- Converts user code into a logical execution plan
- Creates and manages the DAG
- Requests resources from the cluster manager
- Schedules tasks and monitors execution
- Collects results from executor
- Only one driver per application
- Driver failure = application failure

2️⃣ Executors

- Worker processes that run on cluster nodes
- Execute tasks sent by the driver
- Process data partitions in parallel
- Store intermediate data (cache/persist)
- Each executor has fixed CPU cores and memory
- Executors exist for the entire lifetime of the application
- Multiple executors per Spark application

3️⃣ DAG (Directed Acyclic Graph)

- Logical execution plan created by Spark
- Nodes represent transformations
- Edges represent data dependencies
- Created due to lazy evaluation
- No cycles → operations flow in one direction
- Helps Spark optimize execution

4️⃣ DAG → Stages → Tasks

- DAG is divided into stages based on shuffle boundaries
- Each stage is split into tasks
- One task processes one partition
- Tasks are executed in parallel on executors

5️⃣ Lazy Evaluation
- 
- Transformations do not execute immediately
- Execution starts only when an action is called
- Allows Spark to optimize the full DAG

6️⃣ Fault Tolerance

- Spark uses DAG lineage to recompute lost data
- If an executor fails, tasks are re-assigned
- Only failed partitions are recomputed

7️⃣ Performance Optimization

DAG enables:
- Predicate pushdown
- Join reordering
- Stage pipelining
- Minimizes data movement and shuffles

8️⃣ Key Terminology

- Application: One Spark program
- Job: Triggered by an action
- Stage: Group of tasks without shuffle
- Task: Smallest unit of execution

Distirbuted computing:  is the process of spliting a large computational task into smaller subtasks and exectuing them in parallel across multiple machines that work together as a single systems

Spark: Apache spark is the distributed compute engine that abstracts the complexity of parallelism,letting developers focus on business logic

### 1.What is RDD means?
RDD stands for Resilient Distributed Dataset, the fundamental data structure in Apache Spark for big data processing.
​

Core Concept
RDDs represent immutable collections of objects partitioned across cluster nodes for parallel processing. They support fault tolerance by tracking lineage, allowing lost partitions to be recomputed automatically.
​

Key Features
- Resilient: Handles node failures through data lineage reconstruction. (lineage = history/steps --> Spark remembers how an RDD was created, so if data is lost, Spark can rebuild it instead of storing copies.)
- Distributed: Splits data into partitions processed simultaneously on multiple nodes.
- Immutable and Lazy: Transformations create new RDDs; evaluation occurs only on actions like collect() or count().

Operations
- RDDs support transformations (e.g., map, filter) that build computation graphs and actions (e.g., reduce, save) that trigger execution. Data can persist in memory or disk for reuse.

Two types of operations:

🔹 Transformations (lazy)

Create a new RDD from an existing one

Examples:
- map()
- filter()
- flatMap()

👉 No execution happens immediately.

🔹 Actions

Trigger actual computation

Examples:
- count()
- collect()
- saveAsTextFile()

> Example (conceptual)
>
> Data  →  RDD  →  map/filter  →  Action  →  Result


### In-memory computation

Many data processing frameworks—and MapReduce in particular—must perform multiple read or write operations from external storage systems, slowing their performance. RDD helps Apache Spark solve this problem.

RDD reduces disk I/O by using in-memory computation that stores intermediate results from iterative operations in random access memory (RAM). Using in-memory computation and storage can support faster access and near real-time processing.



### 2.Lazy evaluation means? why its used? why this concept introduced? where it will be using ? predecessor and successor of this?

my understanding: delay evaluation like 1 process do then other process will stop until the running function gets completed

- In Spark, all transformations—operations applied to create a new RDD—are “lazy,” which is to say that the data is not loaded or computed immediately.
- Instead, transformations are tracked in a directed acyclic graph (DAG) and run only when there is a specific call to action for a driver program.
- The driver program directs the primary function and operations for cluster computing on Spark jobs, such as aggregation, collecting, counting or saving output to a file system.
- Dozens of possible actions and transformations include aggregateByKey, countByKey, flatMap, groupByKey, reduceByKey and sortbyKey.
- Lazy evaluation helps optimize data processing pipelines by eliminating unnecessary processing and clipping of unnecessary computations.

1️⃣ Lazy evaluation means

Lazy evaluation is a strategy where expressions are evaluated only when their results are required, not when they are defined.

2️⃣ Why is lazy evaluation used?

It improves performance and memory efficiency by avoiding unnecessary computations.

3️⃣ Why was this concept introduced?

It was introduced to optimize resource usage and to support large or infinite data processing.

4️⃣ Where is lazy evaluation used?

It is used in functional programming languages, big data frameworks like Spark, stream processing, and database query execution.

5️⃣ Predecessor of lazy evaluation

The predecessor is eager evaluation, where all expressions are evaluated immediately.

6️⃣ Successor / evolution of lazy evaluation

The successor is hybrid or controlled lazy execution, combining lazy planning with optimized execution (e.g., Spark DAG execution).

## EXAMPLE OF RDD
Scenario: Counting error messages per service from application logs

Sample log data:

- 2026-01-15 serviceA INFO User login
- 2026-01-15 serviceB ERROR Timeout
- 2026-01-15 serviceA ERROR DB failure
- 2026-01-15 serviceC INFO Payment success
- 2026-01-15 serviceB ERROR Connection lost

Step 1️⃣: Create RDD (Loading data)

> logsRDD = sc.textFile("hdfs:///logs/app_logs")


What happens:

- Spark reads the file
- Splits it into partitions across machines
- Creates an RDD object (logsRDD)
- Important: No computation has happened yet.

Think of it like: “I have a book (file) and I’m pointing to it, but I haven’t started reading pages yet.”

Step 2️⃣: Transformation 1 – Filter errors
> errorRDD = logsRDD.filter(lambda line: "ERROR" in line)


- Transformation = a step to define new RDD
- Spark does not execute yet (lazy evaluation)
- It remembers: “Take logsRDD, keep only lines with ERROR”

Result conceptually:
> 2026-01-15 serviceB ERROR Timeout
> 2026-01-15 serviceA ERROR DB failure
> 2026-01-15 serviceB ERROR Connection lost


### Key: Transformation = definition of what to do, not actual calculation.

Step 3️⃣: Transformation 2 – Map to key-value
> serviceErrorRDD = errorRDD.map(lambda line: (line.split()[1], 1))


Splits each line → takes service name → pairs it with 1

Example output (conceptually):
> 
> ("serviceB", 1)
> 
> ("serviceA", 1)
>
> ("serviceB", 1)


Again, no computation yet

Spark just adds this step to the plan

Think of it as: “I’m writing instructions in my notebook on what to do with each line.”

Step 4️⃣: Action – Reduce by key
> errorCount = serviceErrorRDD.reduceByKey(lambda a, b: a + b)
> errorCount.collect()
> 

Action = trigger actual computation

Spark now:
- Reads the log file partitions
- Applies filter() to keep ERROR lines
- Applies map() to create key-value pairs
- Applies reduceByKey() to sum counts per service
- Collects results to driver

Final output:

> serviceA → 1
>
> serviceB → 2


Key: Action = “do all the work defined by transformations and give me the result”
