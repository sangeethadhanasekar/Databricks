## - User-Defined Functions (UDFs)
What are User-Defined Functions (UDFs) in PostgreSQL?

User-Defined Functions (UDFs) are custom functions created by users in PostgreSQL to encapsulate reusable logic that can be called from SQL queries, views, triggers, or applications.

They help in:
- Code reusability
- Cleaner SQL
- Centralized business logic
- Better maintainability

🔹 Languages used to create UDFs

PostgreSQL supports multiple languages:
- SQL
- PL/pgSQL (most common)
- Python (PL/Python)
- JavaScript (PL/V8)
- C

> #Most UDFs are written in PL/pgSQL.

🔹 Basic Syntax (PL/pgSQL)
>     CREATE OR REPLACE FUNCTION function_name(parameters)
>     RETURNS return_datatype
>     LANGUAGE plpgsql
>     AS  $$
>     BEGIN
>       function logic
>       RETURN value;
>     END;
>     $$;

🔹 Example 1: Simple UDF
>     CREATE OR REPLACE FUNCTION add_numbers(a INT, b INT)
>      RETURNS INT
>      LANGUAGE plpgsql
>     AS $$
>     BEGIN
>       RETURN a + b;
>     END;
>     $$;
> 


Calling the function: 

    SELECT add_numbers(10, 20);

🔹 Example 2: UDF with Conditional Logic
>     CREATE OR REPLACE FUNCTION salary_grade(salary NUMERIC)
>     RETURNS TEXT
>     LANGUAGE plpgsql
>     AS $$
>     BEGIN
>       IF salary >= 80000 THEN
>           RETURN 'High';
>       ELSIF salary >= 40000 THEN
>           RETURN 'Medium';
>       ELSE
>           RETURN 'Low';
>       END IF;
>     END;
>     $$;
> 
> SELECT salary_grade(55000);

🔹 Example 3: UDF Returning a Table
>       CREATE OR REPLACE FUNCTION get_employees_by_dept(dept_id INT)
>       RETURNS TABLE(emp_id INT, emp_name TEXT)
>       LANGUAGE plpgsql
>       AS $$
>       BEGIN
>         RETURN QUERY
>         SELECT id, name FROM employees WHERE department_id = dept_id;
>       END;
>       $$;

SELECT * FROM get_employees_by_dept(101);

1️⃣ What is PostgreSQL?

PostgreSQL is an open-source relational database management system (RDBMS) used to store, manage, and retrieve structured data. It provides core database features such as tables, indexes, constraints, transactions, concurrency control, and security. PostgreSQL uses SQL as its primary query language and is widely used in enterprise, analytics, and application backends because of its reliability, extensibility, and standards compliance.

    Example (Template):

    SELECT * FROM employees;

2️⃣ What is PL/pgSQL?

PL/pgSQL is PostgreSQL’s procedural programming language that extends SQL by adding programming constructs such as variables, conditional statements, loops, and exception handling. It is used to write user-defined functions, procedures, and triggers when SQL alone is not sufficient to express complex business logic.

    Example (Template):

    BEGIN
      IF salary > 50000 THEN
          bonus := salary * 0.1;
      END IF;
    END;

3️⃣ Difference Between PostgreSQL and PL/pgSQL

PostgreSQL is the database system that stores data and executes queries, whereas PL/pgSQL is a language that runs inside PostgreSQL to implement procedural logic. PostgreSQL provides the environment and engine, while PL/pgSQL provides control flow and logic capabilities needed for complex operations.

    Example (Template):

    -- PostgreSQL executes this
    SELECT * FROM employees;

    -- PL/pgSQL provides logic for functions
    IF condition THEN
      action;
    END IF;

4️⃣ Why is SQL Alone Not Enough?

SQL is a declarative language, meaning it specifies what data is needed, not how to process it step by step. While SQL is excellent for querying and data manipulation, it lacks features like loops, variables, and error handling. For complex workflows or business rules, SQL alone becomes insufficient.

    Example (Template):

    SELECT name FROM employees WHERE salary > 50000;

5️⃣ Why Do We Use PL/pgSQL (Why “PL”)?

PL stands for Procedural Language, and PL/pgSQL is used because real-world business logic often requires step-by-step execution. PL/pgSQL allows developers to write logic that includes conditions, loops, and exception handling, making it possible to implement validations and calculations directly within the database.

    Example (Template):

    IF total_hours > 40 THEN
      overtime := total_hours - 40;
    END IF;

6️⃣ Can We Create User-Defined Functions Using Only SQL?

Yes, PostgreSQL allows creating SQL-language functions, but they are limited to simple expressions or single queries. These functions cannot contain procedural logic such as loops or conditional branching. SQL functions are best suited for straightforward calculations.

    Example (Template):

    CREATE FUNCTION square(n INT)
    RETURNS INT
    LANGUAGE SQL
    AS $$ SELECT n * n; $$;

7️⃣ When Is PL/pgSQL Mandatory for UDFs?

PL/pgSQL becomes mandatory when a function requires complex logic, such as conditional statements, looping over rows, handling exceptions, or executing multiple SQL statements. These capabilities are not available in plain SQL functions.

    Example (Template):

    CREATE FUNCTION grade(score INT)
    RETURNS TEXT
    LANGUAGE plpgsql
    AS $$
    BEGIN
      IF score >= 60 THEN
          RETURN 'Pass';
      ELSE
          RETURN 'Fail';
      END IF;
    END;
    $$;

    
## - difference between function and Procedure
🔹 Function (PostgreSQL Example)

>       CREATE OR REPLACE FUNCTION get_tax(salary NUMERIC)
>       RETURNS NUMERIC
>       LANGUAGE plpgsql
>       AS $$
>         BEGIN
>             RETURN salary * 0.10;
>         END;
>       $$;


Calling a function:

SELECT get_tax(50000);


- ✔ Used inside queries
- ✔ Returns a value
- ❌ Cannot use COMMIT / ROLLBACK

🔹 Procedure (PostgreSQL Example)

>     CREATE OR REPLACE PROCEDURE update_salary(emp_id INT, increment NUMERIC)
>     LANGUAGE plpgsql
>     AS $$
>     BEGIN
>       UPDATE employees
>       SET salary = salary + increment
>       WHERE id = emp_id;
> 
>       COMMIT;
>     END;
>     $$;


Calling a procedure:

CALL update_salary(101, 5000);


- ✔ Used for business operations
- ✔ Can control transactions
- ❌ Cannot be used inside SELECT

🔹 When to Use What?
✅ Use Function when:
- You need a return value
- Logic is part of a SELECT query
- Doing calculations or validations 
- Writing reusable query logic

✅ Use Procedure when:
- Performing multiple DML operations
- You need COMMIT / ROLLBACK 
- Implementing business workflows
- Running batch jobs

🔑 Very Important Interview Point
- Functions are expression-oriented, procedures are action-oriented.

🔄 Quick Memory Trick
- Function → returns something
- Procedure → performs something

📌 One-Line Interview Answer

“A function returns a value and can be used inside SQL queries, whereas a procedure performs operations, supports transaction control, and is invoked using CALL.”
